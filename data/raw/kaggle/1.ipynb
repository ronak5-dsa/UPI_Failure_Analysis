```python
# Which type of UPI transaction has the highest volume, transaction value, average amount, and failure rate?
print("total transactions count of p2p", p2p_count)
sum_amount_p2p = df[(df['transaction type']=='P2P')]['amount (INR)'].sum()
print("Total Transaction Value of p2p", sum_amount_p2p)
avg_p2p_amount= df[(df['transaction type']=='P2P')]['amount (INR)'].mean()
print("avg  Transaction Value of p2p",  avg_p2p_amount)
p2p_failed_transactions = (
    (df[df['transaction type'] == 'P2P']['transaction_status'] == 'FAILED').sum()
)

failure_rate_p2p = (p2p_failed_transactions / p2p_count) * 100

print("P2P Failed Transactions:", p2p_failed_transactions)
print("P2P Failure Rate:", failure_rate_p2p, "%")
```

```python
# P2M analysis

p2m_count = (df['transaction type'] == 'P2M').sum()

p2m_share = (p2m_count / total_transactions) * 100

p2m_total_value = df.loc[
    df['transaction type'] == 'P2M',
    'amount (INR)'
].sum()

p2m_avg_amount = df.loc[
    df['transaction type'] == 'P2M',
    'amount (INR)'
].mean()

p2m_failed_transactions = (
    (df['transaction type'] == 'P2M') &
    (df['transaction_status'] == 'FAILED')
).sum()

p2m_failure_rate = (
    p2m_failed_transactions / p2m_count
) * 100

print("P2M Transactions:", p2m_count)
print("P2M Share:", p2m_share, "%")
print("P2M Total Transaction Value:", p2m_total_value)
print("P2M Average Transaction Amount:", p2m_avg_amount)
print("P2M Failed Transactions:", p2m_failed_transactions)
print("P2M Failure Rate:", p2m_failure_rate, "%")
```

```python
# Bill Payment analysis

bill_count = (df['transaction type'] == 'Bill Payment').sum()

bill_share = (bill_count / total_transactions) * 100

bill_total_value = df.loc[
    df['transaction type'] == 'Bill Payment',
    'amount (INR)'
].sum()

bill_avg_amount = df.loc[
    df['transaction type'] == 'Bill Payment',
    'amount (INR)'
].mean()

bill_failed_transactions = (
    (df['transaction type'] == 'Bill Payment') &
    (df['transaction_status'] == 'FAILED')
).sum()

bill_failure_rate = (
    bill_failed_transactions / bill_count
) * 100

print("Bill Payment Transactions:", bill_count)
print("Bill Payment Share:", bill_share, "%")
print("Bill Payment Total Transaction Value:", bill_total_value)
print("Bill Payment Average Transaction Amount:", bill_avg_amount)
print("Bill Payment Failed Transactions:", bill_failed_transactions)
print("Bill Payment Failure Rate:", bill_failure_rate, "%")
```

```python
# Recharge analysis

recharge_count = (df['transaction type'] == 'Recharge').sum()

recharge_share = (recharge_count / total_transactions) * 100

recharge_total_value = df.loc[
    df['transaction type'] == 'Recharge',
    'amount (INR)'
].sum()

recharge_avg_amount = df.loc[
    df['transaction type'] == 'Recharge',
    'amount (INR)'
].mean()

recharge_failed_transactions = (
    (df['transaction type'] == 'Recharge') &
    (df['transaction_status'] == 'FAILED')
).sum()

recharge_failure_rate = (
    recharge_failed_transactions / recharge_count
) * 100

print("Recharge Transactions:", recharge_count)
print("Recharge Share:", recharge_share, "%")
print("Recharge Total Transaction Value:", recharge_total_value)
print("Recharge Average Transaction Amount:", recharge_avg_amount)
print("Recharge Failed Transactions:", recharge_failed_transactions)
print("Recharge Failure Rate:", recharge_failure_rate, "%")
```

```python
#"P2P transactions account for the largest share of transaction volume (44.98%) and transaction value (~₹14.72 crore), while recharge transactions show the highest failure rate (5.09%), indicating a potential area for further investigation."
```

```python
## TIMESTAMP ANALYSIS
## HOUR WISE ANALYSIS
print(df['timestamp'].dtype)
df['timestamp']= pd.to_datetime(df['timestamp']) ## object se datetime me convert kr liya
df['hour']= df['timestamp'].dt.hour ## extracted hour from datetime
df['hour'].value_counts().sort_index() ## how many transactions happened in each hour
```

```python
## no. of failed transaction by each hour
failed_transaction_per_hour= df[df['transaction_status'] == 'FAILED'].groupby('hour').size()
print(failed_transaction_per_hour)
```

```python
total_transaction_per_hour = df.groupby('hour').size()
print(total_transaction_per_hour)
```

```python
failure_rate_per_hour = (failed_transaction_per_hour / total_transaction_per_hour) * 100
print(failure_rate_per_hour)
```

```python
## Interestingly, the hour with the highest transaction volume (7 PM) is NOT the hour with the highest failure rate.
## failure rate is:
## 6 AM → 5.40% approximately
```

```python
hourly_analysis = pd.DataFrame({
    'total_transactions': total_transaction_per_hour,
    'failed_transactions': failed_transaction_per_hour,
    'failure_rate': failure_rate_per_hour
})

hourly_analysis
```

```python
df['day']= df['timestamp'].dt.day
df['day'].value_counts().sort_index() # no. of transactions happened each day
```

```python
# failed transactions per day
failed_transactions_per_day = df[(df['transaction_status']=='FAILED')].groupby('day').size()
failed_transactions_per_day
```

```python
total_transactions_per_day= df.groupby('day').size()
total_transactions_per_day
```

```python
failure_rate_per_day= (failed_transactions_per_day/ total_transactions_per_day)*100
failure_rate_per_day
```

```python
daily_analysis = pd.DataFrame({
    'total_transactions':total_transactions_per_day,
    'failed_transactions_per_day':failed_transactions_per_day,
    'failure_rate_per_day':failure_rate_per_day})
daily_analysis
```

```python
# monthly analysis
df['month'] = df['timestamp'].dt.to_period('M')
```
